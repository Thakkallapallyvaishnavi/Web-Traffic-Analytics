```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('website_wata.csv')
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = df.fillna(df.median(numeric_only=True))
```

```python
df['Page Views'].describe()
```

```python
df['Page Views'].hist()
plt.title('Page Views Distribution')
plt.show()
```

```python
df['Session Duration'].describe()
```

```python
df['Session Duration'].hist()
plt.title('Session Duration')
plt.show()
```

```python
df['Bounce Rate'].describe()
```

```python
df['Bounce Rate'].hist()
plt.title('Bounce Rate')
plt.show()
```

```python
df['Traffic Source'].value_counts()
```

```python
df['Traffic Source'].value_counts().plot(kind='bar')
plt.title('Traffic Sources')
plt.show()
```

```python
df['Time on Page'].describe()
```

```python
df['Time on Page'].hist()
plt.title('Time on Page')
plt.show()
```

```python
df['Previous Visits'].describe()
```

```python
df['Previous Visits'].hist()
plt.title('Previous Visits')
plt.show()
```

```python
df['Conversion Rate'].describe()
```

```python
df['Conversion Rate'].hist()
plt.title('Conversion Rate')
plt.show()
```

```python
corr = df.corr(numeric_only=True)
corr
```

```python
plt.figure(figsize=(8,6))
plt.imshow(corr)
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title('Correlation Matrix')
plt.show()
```

Web Traffic Analytics

Key Findings:
1. Page Views indicate user interest levels.
2. Longer Session Duration suggests better engagement.
3. High Bounce Rate indicates user drop-off points.
4. Traffic Sources influence visitor quality.
5. Higher Time on Page improves engagement.
6. Returning visitors show stronger interest.
7. Higher Conversion Rates indicate successful user journeys.

Recommendations:
1. Improve pages with high bounce rates.
2. Increase content quality to improve session duration.
3. Focus marketing on top-performing traffic sources.
4. Encourage repeat visits.
5. Optimize user journeys to improve conversions.
